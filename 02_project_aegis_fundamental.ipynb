```python
def calculate_ai_score(info, price_df):
    """
    AEGIS AI 升級版評分系統
    滿分 100 分，分五個維度
    """
    score = 0

    # ===== 1. 獲利能力 (30分) =====
    roe = info.get("returnOnEquity") or 0
    gross_margin = info.get("grossMargins") or 0
    profit_margin = info.get("profitMargins") or 0

    if roe > 0.20:
        score += 15
    elif roe > 0.10:
        score += 8

    if gross_margin > 0.50:
        score += 10
    elif gross_margin > 0.30:
        score += 5

    if profit_margin > 0.20:
        score += 5
    elif profit_margin > 0.10:
        score += 3

    # ===== 2. 成長力 (25分) =====
    revenue_growth = info.get("revenueGrowth") or 0
    earnings_growth = info.get("earningsGrowth") or 0

    if revenue_growth > 0.30:
        score += 15
    elif revenue_growth > 0.10:
        score += 8
    elif revenue_growth > 0:
        score += 3

    if earnings_growth > 0.30:
        score += 10
    elif earnings_growth > 0.10:
        score += 5

    # ===== 3. 財務健康 (20分) =====
    debt_ratio = info.get("debtToEquity") or 0
    current_ratio = info.get("currentRatio") or 0

    if debt_ratio < 50:
        score += 10
    elif debt_ratio < 100:
        score += 5

    if current_ratio > 2:
        score += 10
    elif current_ratio > 1:
        score += 5

    # ===== 4. 估值 (10分) =====
    pe = info.get("forwardPE") or info.get("trailingPE") or 0

    if 0 < pe < 15:
        score += 10
    elif 15 <= pe < 25:
        score += 6
    elif 25 <= pe < 40:
        score += 3

    # ===== 5. 技術面 (15分) =====
    if len(price_df) >= 60:
        close = price_df["Close"]
        ma20 = close.rolling(20).mean().iloc[-1]
        ma60 = close.rolling(60).mean().iloc[-1]
        latest = close.iloc[-1]

        if latest > ma20 > ma60:
            score += 10
        elif latest > ma20:
            score += 5

        delta = close.diff()
        rs = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
        rsi = (100 - (100 / (1 + rs))).iloc[-1]

        if 40 < rsi < 70:
            score += 5

    return min(score, 100)

print("✅ AEGIS AI 升級版評分函式載入完成")
```

```python
import yfinance as yf
import pandas as pd
import datetime

stock_list = [
    "2330.TW", "2308.TW", "2317.TW", "2454.TW", "3008.TW",
    "2376.TW", "3231.TW", "2356.TW", "2881.TW", "2882.TW",
    "2303.TW", "3037.TW", "2382.TW", "6415.TW", "3443.TW"
]

def aegis_daily_run():
    results = []
    now = __import__('datetime').datetime.now().strftime("%Y-%m-%d %H:%M")

    for stock_id in stock_list:
        try:
            ticker = yf.Ticker(stock_id)
            df = ticker.history(period="6mo")

            if df.empty or len(df) < 60:
                continue

            close = df["Close"]
            volume = df["Volume"]

            current_price = close.iloc[-1]
            ma20 = close.rolling(20).mean().iloc[-1]
            ma60 = close.rolling(60).mean().iloc[-1]
            avg_volume = volume.rolling(20).mean().iloc[-1]
            today_volume = volume.iloc[-1]

            score = 0
            reasons = []

            if current_price > ma20:
                score += 20
                reasons.append("股價站上 MA20")

            if ma20 > ma60:
                score += 20
                reasons.append("MA20 高於 MA60")

            if today_volume > avg_volume * 1.5:
                score += 25
                reasons.append("成交量放大")

            resistance = close.tail(60).max()
            support = close.tail(60).min()

            if current_price > resistance * 0.98:
                score += 25
                reasons.append("接近壓力區")

            if current_price > close.iloc[-2]:
                score += 10
                reasons.append("今日收盤高於昨日")

            results.append({
                "時間": now,
                "股票代號": stock_id,
                "目前股價": round(current_price, 2),
                "MA20": round(ma20, 2),
                "MA60": round(ma60, 2),
                "支撐": round(support, 2),
                "壓力": round(resistance, 2),
                "AI分數": score,
                "原因": "；".join(reasons)
            })

        except Exception as e:
            print(stock_id, "失敗：", e)

    ranking_df = pd.DataFrame(results)

    if ranking_df.empty:
        send_line_message("AEGIS AI：目前沒有成功取得股票資料")
        return

    ranking_df = ranking_df.sort_values("AI分數", ascending=False)
    top_stock = ranking_df.iloc[0]

    ranking_df.to_excel("AEGIS_AI_每日戰報_最新.xlsx", index=False)

    message = f"""
AEGIS AI 即時股票戰報

時間：{now}

今日最強股票：
{top_stock["股票代號"]}

AI 分數：{top_stock["AI分數"]}

目前股價：{top_stock["目前股價"]}
MA20：{top_stock["MA20"]}
MA60：{top_stock["MA60"]}

支撐：{top_stock["支撐"]}
壓力：{top_stock["壓力"]}

原因：
{top_stock["原因"]}
"""

    send_line_message(message)
    print("AEGIS AI 真實戰報已送出")
    print(ranking_df.head(10))
```

```python
import yfinance as yf
import pandas as pd

def backtest_aegis(stock_list, period="1y"):
    records = []

    for stock_id in stock_list:
        try:
            df = yf.Ticker(stock_id).history(period=period)

            if df.empty or len(df) < 70:
                continue

            df["MA20"] = df["Close"].rolling(20).mean()
            df["MA60"] = df["Close"].rolling(60).mean()
            df["AVG_VOL20"] = df["Volume"].rolling(20).mean()

            for i in range(60, len(df) - 5):
                score = 0
                reasons = []

                current_price = df["Close"].iloc[i]
                yesterday_price = df["Close"].iloc[i - 1]
                ma20 = df["MA20"].iloc[i]
                ma60 = df["MA60"].iloc[i]
                today_volume = df["Volume"].iloc[i]
                avg_volume = df["AVG_VOL20"].iloc[i]
                resistance = df["Close"].iloc[i-60:i].max()

                if current_price > ma20:
                    score += 20
                    reasons.append("站上MA20")

                if ma20 > ma60:
                    score += 20
                    reasons.append("MA20高於MA60")

                if today_volume > avg_volume * 1.5:
                    score += 25
                    reasons.append("成交量放大")

                if current_price > resistance * 0.98:
                    score += 25
                    reasons.append("接近壓力")

                if current_price > yesterday_price:
                    score += 10
                    reasons.append("收盤高於昨日")

                next_1_return = (df["Close"].iloc[i+1] - current_price) / current_price * 100
                next_5_return = (df["Close"].iloc[i+5] - current_price) / current_price * 100

                records.append({
                    "股票代號": stock_id,
                    "日期": df.index[i].strftime("%Y-%m-%d"),
                    "AI分數": score,
                    "隔日報酬%": round(next_1_return, 2),
                    "5日報酬%": round(next_5_return, 2),
                    "原因": "；".join(reasons)
                })

        except Exception as e:
            print(stock_id, "回測失敗:", e)

    backtest_df = pd.DataFrame(records)
    return backtest_df
```

```python
def add_advanced_features(df):

    df = df.copy()

    df["MA20"] = df["Close"].rolling(20).mean()
    df["MA60"] = df["Close"].rolling(60).mean()

    df["成交量MA20"] = df["Volume"].rolling(20).mean()

    df["成交量倍率"] = (
        df["Volume"] / df["成交量MA20"]
    )

    df["報酬率"] = df["Close"].pct_change()

    df["波動率"] = (
        df["報酬率"].rolling(20).std()
    )

    df["20日最高"] = (
        df["High"].rolling(20).max()
    )

    df["20日最低"] = (
        df["Low"].rolling(20).min()
    )

    df["距離壓力"] = (
        (df["20日最高"] - df["Close"])
        / df["Close"]
    )

    df["距離支撐"] = (
        (df["Close"] - df["20日最低"])
        / df["Close"]
    )

    df["多頭排列"] = (
        df["MA20"] > df["MA60"]
    ).astype(int)

    return df
```

```python
def get_us_market_status():
    nasdaq = yf.Ticker("^IXIC").history(period="6mo")
    sox = yf.Ticker("^SOX").history(period="6mo")
    vix = yf.Ticker("^VIX").history(period="1mo")

    nasdaq_close = nasdaq["Close"]
    sox_close = sox["Close"]
    vix_close = vix["Close"]

    nasdaq_now = nasdaq_close.iloc[-1]
    nasdaq_ma20 = nasdaq_close.rolling(20).mean().iloc[-1]

    sox_now = sox_close.iloc[-1]
    sox_ma20 = sox_close.rolling(20).mean().iloc[-1]

    vix_now = vix_close.iloc[-1]

    score = 0
    reason = []

    if nasdaq_now > nasdaq_ma20:
        score += 30
        reason.append("NASDAQ 短線偏強，科技股環境較有利")
    else:
        reason.append("NASDAQ 短線偏弱，科技股需保守")

    if sox_now > sox_ma20:
        score += 40
        reason.append("費半指數偏強，對台灣半導體族群有利")
    else:
        reason.append("費半指數偏弱，半導體族群需留意風險")

    if vix_now < 20:
        score += 30
        reason.append("VIX 恐慌指數偏低，市場風險情緒穩定")
    else:
        reason.append("VIX 偏高，市場避險情緒較強")

    if score >= 70:
        decision = "美股環境偏多，有利台股強勢股表現"
    elif score >= 40:
        decision = "美股環境普通，適合保守觀察"
    else:
        decision = "美股環境偏弱，建議降低風險"

    return score, reason, decision
```

```python
def enrich_stock_row(stock_id):
    df = yf.Ticker(stock_id).history(period="6mo")
    if df.empty or len(df) < 60:
        return None

    close = df["Close"]
    volume = df["Volume"]

    price = close.iloc[-1]
    ma20 = close.rolling(20).mean().iloc[-1]
    ma60 = close.rolling(60).mean().iloc[-1]
    support = close.tail(60).min()
    resistance = close.tail(60).max()
    avg_vol = volume.rolling(20).mean().iloc[-1]
    vol_ratio = volume.iloc[-1] / avg_vol

    reasons = []
    if price > ma20:
        reasons.append("股價高於20日均線，短線偏強")
    if ma20 > ma60:
        reasons.append("20日均線高於60日均線，中期趨勢偏多")
    if vol_ratio > 1.5:
        reasons.append("成交量明顯放大，資金關注度提高")
    if price > resistance * 0.98:
        reasons.append("股價接近壓力區，若突破可能轉強")

    return {
        "目前股價": round(price, 2),
        "MA20": round(ma20, 2),
        "MA60": round(ma60, 2),
        "支撐": round(support, 2),
        "壓力": round(resistance, 2),
        "目標價": round(price * 1.12, 2),
        "停損價": round(price * 0.93, 2),
        "成交量倍率": round(vol_ratio, 2),
        "分析理由": "；".join(reasons)
    }
```

```python
def build_ai_reason(row, info):
    reasons = []

    if info["目前股價"] > info["MA20"]:
        reasons.append("股價站上20日均線，短線結構偏強")

    if info["MA20"] > info["MA60"]:
        reasons.append("20日均線高於60日均線，中期趨勢偏多")

    if info["成交量倍率"] > 1.5:
        reasons.append(f"成交量放大至 {info['成交量倍率']} 倍，資金關注度提高")

    if hasattr(row, "roe") and row.roe > 0.3:
        reasons.append("ROE表現優於市場平均，基本面效率較佳")

    if hasattr(row, "growth") and row.growth > 0.5:
        reasons.append("營收或成長動能偏強")

    if hasattr(row, "RSI"):
        if row.RSI < 30:
            reasons.append("RSI偏低，短線可能有反彈機會")
        elif row.RSI > 70:
            reasons.append("RSI偏高，需留意短線過熱")
        else:
            reasons.append("RSI位於中性區間，尚未明顯過熱")

    if us_score >= 70:
        reasons.append("美股環境偏多，NASDAQ、費半與VIX對科技股較有利")
    elif us_score < 40:
        reasons.append("美股環境偏弱，需降低追高風險")

    if news_result["score"] >= 70:
        reasons.append("新聞情緒偏正向")
    elif news_result["score"] <= 40:
        reasons.append("新聞情緒偏負面，需保守")

    if risk_score >= 75:
        reasons.append("整體市場風險偏高，不宜重壓")
    elif risk_score >= 40:
        reasons.append("市場風險中等，適合控制部位")
    else:
        reasons.append("市場風險目前可接受")

    if len(reasons) == 0:
        reasons.append("目前資料不足，暫以技術分數作為觀察依據")

    return "；".join(reasons)

```

```python
def auto_repair_model(backtest_df):
    results = []

    for threshold in range(40, 91, 5):
        signal = backtest_df[backtest_df["AI分數"] >= threshold]

        if len(signal) < 30:
            continue

        win_rate_5d = (signal["5日報酬%"] > 0).mean() * 100
        avg_return_5d = signal["5日報酬%"].mean()

        results.append({
            "門檻": threshold,
            "訊號數": len(signal),
            "5日勝率": round(win_rate_5d, 2),
            "平均5日報酬": round(avg_return_5d, 2)
        })

    repair_df = pd.DataFrame(results)

    best = repair_df.sort_values(
        ["平均5日報酬", "5日勝率"],
        ascending=False
    ).iloc[0]

    print("===== AEGIS AI 自動修復結果 =====")
    print(repair_df)
    print("\n最佳AI門檻：", best["門檻"])
    print("5日勝率：", best["5日勝率"], "%")
    print("平均5日報酬：", best["平均5日報酬"], "%")

    return best, repair_df
```

```python
import requests
from dotenv import load_dotenv
import os

load_dotenv()
NEWS_API_KEY = os.environ.get("NEWS_API_KEY", "")

def get_stock_news(keyword, max_articles=10):
    """抓真實新聞並分析情緒"""

    url = "https://newsapi.org/v2/everything"
    params = {
        "q": keyword,
        "language": "zh",
        "sortBy": "publishedAt",
        "pageSize": max_articles,
        "apiKey": NEWS_API_KEY
    }

    response = requests.get(url, params=params)
    data = response.json()

    if data.get("status") != "ok":
        # 英文備援
        params["language"] = "en"
        response = requests.get(url, params=params)
        data = response.json()

    articles = data.get("articles", [])

    positive_words = ["surge", "growth", "beat", "record", "rally",
                      "上漲", "創新高", "成長", "利多", "突破", "大增"]
    negative_words = ["crash", "fall", "miss", "layoff", "recession",
                      "下跌", "衰退", "虧損", "利空", "暴跌", "裁員"]

    total_score = 0
    results = []

    for article in articles:
        title = article.get("title", "")
        score = 0

        for word in positive_words:
            if word.lower() in title.lower():
                score += 1

        for word in negative_words:
            if word.lower() in title.lower():
                score -= 1

        total_score += score
        results.append({"title": title, "score": score})

    sentiment = "偏多 📈" if total_score > 0 else "偏空 📉" if total_score < 0 else "中性 ➡️"

    return {
        "keyword": keyword,
        "total_score": total_score,
        "sentiment": sentiment,
        "articles": results
    }

print("✅ 新聞情緒分析函式載入完成")
```

```python
def get_technical_score(price_df):
    """
    技術面綜合評分
    滿分 100 分
    """
    if len(price_df) < 60:
        return 0, []

    score = 0
    signals = []
    close = price_df["Close"]
    volume = price_df["Volume"]

    # ===== 1. 均線系統 (30分) =====
    ma20 = close.rolling(20).mean().iloc[-1]
    ma60 = close.rolling(60).mean().iloc[-1]
    latest = close.iloc[-1]

    if latest > ma20 > ma60:
        score += 30
        signals.append("多頭排列 ✅")
    elif latest > ma20:
        score += 15
        signals.append("短線偏多")
    elif latest < ma20 < ma60:
        signals.append("空頭排列 ⚠️")

    # ===== 2. RSI (20分) =====
    delta = close.diff()
    rs = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
    rsi = (100 - (100 / (1 + rs))).iloc[-1]

    if 50 < rsi < 70:
        score += 20
        signals.append(f"RSI強勢區 {round(rsi,1)}")
    elif 40 < rsi <= 50:
        score += 10
        signals.append(f"RSI中性 {round(rsi,1)}")
    elif rsi >= 70:
        signals.append(f"RSI過熱 {round(rsi,1)} ⚠️")
    elif rsi <= 30:
        score += 15
        signals.append(f"RSI超跌反彈機會 {round(rsi,1)}")

    # ===== 3. MACD (25分) =====
    ema12 = close.ewm(span=12, adjust=False).mean()
    ema26 = close.ewm(span=26, adjust=False).mean()
    macd = ema12 - ema26
    signal_line = macd.ewm(span=9, adjust=False).mean()

    macd_now = macd.iloc[-1]
    macd_prev = macd.iloc[-2]
    signal_now = signal_line.iloc[-1]
    signal_prev = signal_line.iloc[-2]

    # 黃金交叉
    if macd_prev < signal_prev and macd_now > signal_now:
        score += 25
        signals.append("MACD黃金交叉 🔥")
    elif macd_now > signal_now:
        score += 15
        signals.append("MACD多頭")
    else:
        signals.append("MACD空頭")

    # ===== 4. 布林通道 (15分) =====
    bb_mid = close.rolling(20).mean()
    bb_std = close.rolling(20).std()
    bb_upper = bb_mid + 2 * bb_std
    bb_lower = bb_mid - 2 * bb_std

    bb_pos = (latest - bb_lower.iloc[-1]) / (bb_upper.iloc[-1] - bb_lower.iloc[-1])

    if 0.5 < bb_pos < 0.8:
        score += 15
        signals.append("布林中上軌強勢")
    elif bb_pos >= 0.8:
        signals.append("布林上軌壓力區")
    elif bb_pos <= 0.2:
        score += 10
        signals.append("布林下軌支撐反彈")

    # ===== 5. 成交量 (10分) =====
    avg_vol = volume.rolling(20).mean().iloc[-1]
    today_vol = volume.iloc[-1]

    if today_vol > avg_vol * 1.5:
        score += 10
        signals.append("成交量放大 🔥")
    elif today_vol > avg_vol * 1.2:
        score += 5
        signals.append("成交量偏大")

    return min(score, 100), signals

print("✅ 技術面升級評分函式載入完成")
```

```python
# ===== AEGIS AI 三維整合評分 =====
def aegis_full_score(stock_id, keyword=None):
    """
    三維整合：基本面 + 技術面 + 新聞情緒
    """
    try:
        ticker = yf.Ticker(stock_id)
        info = ticker.info
        price_df = ticker.history(period="6mo")

        if price_df.empty or len(price_df) < 60:
            return None

        # 基本面 (50%)
        fundamental = calculate_ai_score(info, price_df)

        # 技術面 (35%)
        technical, signals = get_technical_score(price_df)

        # 新聞情緒 (15%)
        if keyword:
            news = get_stock_news(keyword)
            news_score = min(max(50 + news["total_score"] * 10, 0), 100)
            sentiment = news["sentiment"]
        else:
            news_score = 50
            sentiment = "未抓取"

        # 加權總分
        total = round(
            fundamental * 0.50 +
            technical   * 0.35 +
            news_score  * 0.15,
            1
        )

        return {
            "stock": stock_id,
            "total": total,
            "fundamental": fundamental,
            "technical": technical,
            "news_score": news_score,
            "sentiment": sentiment,
            "signals": signals
        }

    except Exception as e:
        print(f"{stock_id} 失敗: {e}")
        return None

print("✅ 三維整合評分函式載入完成")
```

```python
def run_backtest(stock_id, period="1y", min_score=60):
    """
    升級版回測：用三維評分當買入條件
    每次買入持有5天，統計勝率和報酬
    """
    import yfinance as yf

    ticker = yf.Ticker(stock_id)
    df = ticker.history(period=period)

    if df.empty or len(df) < 80:
        print(f"{stock_id} 資料不足")
        return None

    df["MA20"] = df["Close"].rolling(20).mean()
    df["MA60"] = df["Close"].rolling(60).mean()

    delta = df["Close"].diff()
    rs = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
    df["RSI"] = 100 - (100 / (1 + rs))

    ema12 = df["Close"].ewm(span=12, adjust=False).mean()
    ema26 = df["Close"].ewm(span=26, adjust=False).mean()
    df["MACD"] = ema12 - ema26
    df["Signal"] = df["MACD"].ewm(span=9, adjust=False).mean()

    trades = []

    for i in range(60, len(df) - 5):
        close  = df["Close"].iloc[i]
        ma20   = df["MA20"].iloc[i]
        ma60   = df["MA60"].iloc[i]
        rsi    = df["RSI"].iloc[i]
        macd   = df["MACD"].iloc[i]
        signal = df["Signal"].iloc[i]
        macd_prev   = df["MACD"].iloc[i-1]
        signal_prev = df["Signal"].iloc[i-1]

        # 買入條件
        score = 0
        if close > ma20 > ma60:
            score += 30
        if 40 < rsi < 70:
            score += 20
        if macd_prev < signal_prev and macd > signal:
            score += 30   # 黃金交叉
        elif macd > signal:
            score += 15

        if score < min_score:
            continue

        # 5日後報酬
        future_price = df["Close"].iloc[i + 5]
        ret = (future_price - close) / close * 100

        trades.append({
            "date":  df.index[i].strftime("%Y-%m-%d"),
            "buy":   round(close, 2),
            "sell":  round(future_price, 2),
            "ret":   round(ret, 2),
            "score": score
        })

    if not trades:
        print(f"{stock_id} 沒有符合條件的交易")
        return None

    import pandas as pd
    bt = pd.DataFrame(trades)

    win_rate   = (bt["ret"] > 0).mean() * 100
    avg_ret    = bt["ret"].mean()
    best       = bt["ret"].max()
    worst      = bt["ret"].min()
    total      = len(bt)

    print(f"\n===== AEGIS AI 回測結果：{stock_id} =====")
    print(f"回測期間：{period}")
    print(f"交易次數：{total}")
    print(f"勝率：    {round(win_rate, 1)} %")
    print(f"平均報酬：{round(avg_ret, 2)} %")
    print(f"最佳交易：{best} %")
    print(f"最差交易：{worst} %")
    print(f"\n近5筆交易：")
    print(bt.tail(5)[["date","buy","sell","ret","score"]].to_string(index=False))

    return bt

print("✅ 升級版回測函式載入完成")
```

```python
def sector_rotation_radar():
    """
    產業輪動雷達
    判斷今天資金集中在哪個產業
    """
    sectors = {
        "半導體": semiconductor,
        "AI電子": ai_electronics,
        "金融":   financials,
        "航運":   shipping,
        "傳產":   traditional,
    }

    print("===== AEGIS AI 產業輪動雷達 =====\n")

    sector_scores = []

    for sector_name, stock_dict in sectors.items():
        print(f"掃描 {sector_name}...", end=" ")

        # 技術面平均分
        tech_scores = []
        strong_stocks = []

        for stock_id in list(stock_dict.keys())[:10]:  # 每產業取前10檔
            try:
                df = yf.Ticker(stock_id).history(period="3mo")
                if df.empty or len(df) < 60:
                    continue

                tech, signals = get_technical_score(df)
                tech_scores.append(tech)

                if tech >= 60:
                    strong_stocks.append(stock_id)

            except:
                continue

        if not tech_scores:
            print("資料不足")
            continue

        avg_tech = round(sum(tech_scores) / len(tech_scores), 1)

        # 產業新聞分數
        news = get_sector_news_score(sector_name)
        news_score = news["score"]
        sentiment = news["sentiment"]

        # 產業總分
        total = round(avg_tech * 0.6 + news_score * 0.4, 1)

        sector_scores.append({
            "sector":        sector_name,
            "total":         total,
            "tech_avg":      avg_tech,
            "news_score":    news_score,
            "sentiment":     sentiment,
            "strong_stocks": strong_stocks,
        })

        print(f"總分 {total}　強勢股數：{len(strong_stocks)}")

    # 排行
    sector_scores.sort(key=lambda x: x["total"], reverse=True)

    print("\n===== 產業輪動排行 =====")
    print(f"{'排名':<4} {'產業':<8} {'總分':<6} {'技術':<6} {'新聞':<6} {'情緒'}")
    print("-" * 50)

    for i, s in enumerate(sector_scores, 1):
        medal = "🥇" if i == 1 else "🥈" if i == 2 else "🥉" if i == 3 else "  "
        print(f"{medal} {i}  {s['sector']:<8} {s['total']:<6} {s['tech_avg']:<6} {s['news_score']:<6} {s['sentiment']}")

    print("\n===== 今日主線產業 =====")
    top = sector_scores[0]
    print(f"🔥 資金最集中：{top['sector']}")
    print(f"   產業總分：{top['total']}")
    print(f"   新聞情緒：{top['sentiment']}")
    print(f"   強勢個股：{' | '.join(top['strong_stocks'][:5])}")

    return sector_scores

print("✅ 產業輪動雷達載入完成")
```

```python
def intraday_alert(watchlist_ids, check_interval_min=10):
    """
    盤中異動警報
    偵測：突破MA20、成交量暴增、RSI超買超賣
    """
    import datetime as dt

    print(f"===== AEGIS AI 盤中監控啟動 =====")
    print(f"監控股數：{len(watchlist_ids)}")
    print(f"檢查頻率：每 {check_interval_min} 分鐘")
    print(f"啟動時間：{dt.datetime.now().strftime('%H:%M')}\n")

    alerts = []

    for stock_id in watchlist_ids:
        try:
            df = yf.Ticker(stock_id).history(period="3mo")
            if df.empty or len(df) < 20:
                continue

            close   = df["Close"]
            volume  = df["Volume"]
            ma20    = close.rolling(20).mean()
            avg_vol = volume.rolling(20).mean()

            latest      = close.iloc[-1]
            prev        = close.iloc[-2]
            ma20_now    = ma20.iloc[-1]
            vol_now     = volume.iloc[-1]
            avg_vol_now = avg_vol.iloc[-1]

            # RSI
            delta = close.diff()
            rs    = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
            rsi   = (100 - (100 / (1 + rs))).iloc[-1]

            triggered = []
            level = 0

            # 突破 MA20
            if prev < ma20.iloc[-2] and latest > ma20_now:
                triggered.append("🚀 突破MA20")
                level += 3

            # 成交量暴增 2倍
            if vol_now > avg_vol_now * 2:
                triggered.append("💥 成交量暴增2倍")
                level += 3

            # 成交量放大 1.5倍
            elif vol_now > avg_vol_now * 1.5:
                triggered.append("📈 成交量放大")
                level += 1

            # RSI 超賣反彈
            if rsi < 30:
                triggered.append(f"⚡ RSI超跌({round(rsi,1)}) 反彈機會")
                level += 2

            # RSI 強勢區
            elif 55 < rsi < 70:
                triggered.append(f"✅ RSI強勢區({round(rsi,1)})")
                level += 1

            # 今日漲幅 > 3%
            change_pct = (latest - prev) / prev * 100
            if change_pct > 3:
                triggered.append(f"🔥 今日漲幅 {round(change_pct,1)}%")
                level += 2

            if triggered:
                alerts.append({
                    "stock":   stock_id,
                    "price":   round(latest, 2),
                    "change":  round(change_pct, 2),
                    "rsi":     round(rsi, 1),
                    "level":   level,
                    "reasons": triggered,
                })

        except Exception as e:
            continue

    # 按強度排序
    alerts.sort(key=lambda x: x["level"], reverse=True)

    if not alerts:
        print("目前無異動警報，市場平靜")
        return []

    print(f"⚠️  偵測到 {len(alerts)} 檔異動！\n")

    message = f"AEGIS AI 盤中異動警報\n{dt.datetime.now().strftime('%H:%M')}\n\n"

    for a in alerts:
        stars = "🔴" * min(a["level"], 5)
        print(f"{stars} {a['stock']}")
        print(f"   股價：{a['price']}　漲跌：{a['change']:+.2f}%　RSI：{a['rsi']}")
        for r in a["reasons"]:
            print(f"   {r}")
        print()

        message += f"{stars} {a['stock']} {a['price']}\n"
        message += f"漲跌：{a['change']:+.2f}%\n"
        for r in a["reasons"]:
            message += f"{r}\n"
        message += "\n"

    # LINE 推播
    send_line_message(message)
    print("✅ 盤中警報已推播 LINE！")

    return alerts

print("✅ 盤中異動警報載入完成")
```

```python
import json
import os
import datetime as dt

LEARNING_FILE = "AEGIS_learning.json"

def save_recommendation(stock_id, price, score, signals, reason="三維評分"):
    """
    儲存每次推薦記錄
    5天後自動驗證結果
    """
    today = dt.datetime.now().strftime("%Y-%m-%d")
    verify_date = (dt.datetime.now() + dt.timedelta(days=5)).strftime("%Y-%m-%d")

    record = {
        "date":        today,
        "stock":       stock_id,
        "buy_price":   price,
        "score":       score,
        "signals":     signals,
        "reason":      reason,
        "verify_date": verify_date,
        "verified":    False,
        "result":      None
    }

    # 讀取現有記錄
    if os.path.exists(LEARNING_FILE):
        with open(LEARNING_FILE, "r", encoding="utf-8") as f:
            data = json.load(f)
    else:
        data = []

    data.append(record)

    with open(LEARNING_FILE, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"✅ 已記錄推薦：{stock_id} @ {price}，{verify_date} 驗證")


def verify_past_recommendations():
    """
    驗證過去的推薦是否正確
    自動更新勝率統計
    """
    if not os.path.exists(LEARNING_FILE):
        print("尚無推薦記錄")
        return

    with open(LEARNING_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    today = dt.datetime.now().strftime("%Y-%m-%d")
    updated = False

    for record in data:
        if record["verified"]:
            continue
        if record["verify_date"] > today:
            continue

        # 到期了，去抓現在股價
        try:
            ticker = yf.Ticker(record["stock"])
            current = ticker.history(period="5d")["Close"].iloc[-1]
            buy     = record["buy_price"]
            ret     = round((current - buy) / buy * 100, 2)

            record["sell_price"] = round(current, 2)
            record["return_pct"] = ret
            record["result"]     = "✅ 獲利" if ret > 0 else "❌ 虧損"
            record["verified"]   = True
            updated = True

            print(f"{record['stock']} 買@{buy} 現@{round(current,2)} 報酬:{ret:+.2f}% {record['result']}")

        except Exception as e:
            print(f"{record['stock']} 驗證失敗：{e}")

    if updated:
        with open(LEARNING_FILE, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

    # 統計勝率
    verified = [r for r in data if r["verified"]]
    if verified:
        wins    = sum(1 for r in verified if r["return_pct"] > 0)
        avg_ret = sum(r["return_pct"] for r in verified) / len(verified)
        print(f"\n===== AEGIS AI 學習統計 =====")
        print(f"總推薦次數：{len(data)}")
        print(f"已驗證：{len(verified)}")
        print(f"勝率：{round(wins/len(verified)*100, 1)} %")
        print(f"平均報酬：{round(avg_ret, 2)} %")
    else:
        print("尚無已驗證記錄，繼續累積中...")


def show_learning_history():
    """顯示所有推薦歷史"""
    if not os.path.exists(LEARNING_FILE):
        print("尚無記錄")
        return

    with open(LEARNING_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    print(f"===== AEGIS AI 推薦歷史 ({len(data)}筆) =====")
    for r in data[-10:]:  # 最近10筆
        status = r["result"] if r["verified"] else f"待驗證({r['verify_date']})"
        ret    = f"{r['return_pct']:+.2f}%" if r.get("return_pct") else "---"
        print(f"{r['date']} {r['stock']:10} 買@{r['buy_price']} 分數:{r['score']} {ret} {status}")

print("✅ AI 學習記錄系統載入完成")
```

```python
import json, os
import datetime as dt

WATCHLIST_FILE = "AEGIS_watchlist.json"

def add_to_watchlist(stock_id, buy_price, stop_loss_pct=0.07, target_pct=0.15):
    """
    加入追蹤清單
    stop_loss_pct: 停損百分比（預設7%）
    target_pct: 目標百分比（預設15%）
    """
    name = get_name(stock_id)
    stop_loss  = round(buy_price * (1 - stop_loss_pct), 2)
    target     = round(buy_price * (1 + target_pct), 2)
    date_added = dt.datetime.now().strftime("%Y-%m-%d %H:%M")

    record = {
        "stock":      stock_id,
        "name":       name,
        "buy_price":  buy_price,
        "stop_loss":  stop_loss,
        "target":     target,
        "date_added": date_added,
        "status":     "持有中",
        "hit":        None
    }

    if os.path.exists(WATCHLIST_FILE):
        with open(WATCHLIST_FILE, "r", encoding="utf-8") as f:
            data = json.load(f)
    else:
        data = []

    # 避免重複加入
    data = [d for d in data if d["stock"] != stock_id]
    data.append(record)

    with open(WATCHLIST_FILE, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)

    print(f"✅ 已加入追蹤：【{name}】{stock_id}")
    print(f"   買入價：{buy_price}")
    print(f"   停損價：{stop_loss}（-{stop_loss_pct*100:.0f}%）")
    print(f"   目標價：{target}（+{target_pct*100:.0f}%）")


def check_stop_loss():
    """
    檢查所有追蹤股票
    跌破停損或達到目標價立刻警報
    """
    if not os.path.exists(WATCHLIST_FILE):
        print("追蹤清單是空的")
        return

    with open(WATCHLIST_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    active = [d for d in data if d["status"] == "持有中"]

    if not active:
        print("目前沒有持有中的追蹤股票")
        return

    print(f"===== AEGIS AI 停損警報檢查 =====")
    print(f"追蹤中：{len(active)} 檔\n")

    alerts = []
    updated = False

    for record in data:
        if record["status"] != "持有中":
            continue

        try:
            df = yf.Ticker(record["stock"]).history(period="2d")
            current = round(float(df["Close"].iloc[-1]), 2)
            buy     = record["buy_price"]
            ret     = round((current - buy) / buy * 100, 2)
            name    = record["name"]

            status_icon = "🟢" if current > buy else "🔴"
            print(f"{status_icon} 【{name}】{record['stock']}")
            print(f"   買入：{buy}　現價：{current}　報酬：{ret:+.2f}%")
            print(f"   停損：{record['stop_loss']}　目標：{record['target']}")

            # 停損觸發
            if current <= record["stop_loss"]:
                msg = f"🚨 停損警報！\n【{name}】{record['stock']}\n現價：{current}\n已跌破停損價 {record['stop_loss']}\n報酬：{ret:+.2f}%\n建議立即停損出場！"
                alerts.append(msg)
                record["status"] = "已停損"
                record["hit"]    = f"{current} ({dt.datetime.now().strftime('%Y-%m-%d')})"
                updated = True
                print(f"   🚨 停損觸發！")

            # 目標達成
            elif current >= record["target"]:
                msg = f"🎯 目標達成！\n【{name}】{record['stock']}\n現價：{current}\n已達目標價 {record['target']}\n報酬：{ret:+.2f}%\n可考慮獲利了結！"
                alerts.append(msg)
                record["status"] = "目標達成"
                record["hit"]    = f"{current} ({dt.datetime.now().strftime('%Y-%m-%d')})"
                updated = True
                print(f"   🎯 目標達成！")

            print()

        except Exception as e:
            print(f"   ❌ 無法取得資料：{e}\n")

    if updated:
        with open(WATCHLIST_FILE, "w", encoding="utf-8") as f:
            json.dump(data, f, ensure_ascii=False, indent=2)

    # 推播警報
    for msg in alerts:
        send_line_message(msg)
        print(f"已推播：{msg[:30]}...")

    if not alerts:
        print("✅ 所有持股正常，無停損或目標觸發")


def show_watchlist():
    """顯示追蹤清單"""
    if not os.path.exists(WATCHLIST_FILE):
        print("追蹤清單是空的")
        return

    with open(WATCHLIST_FILE, "r", encoding="utf-8") as f:
        data = json.load(f)

    print(f"===== AEGIS AI 追蹤清單 ({len(data)}檔) =====")
    for r in data:
        icon = "🟢" if r["status"] == "持有中" else "⚫"
        print(f"{icon} 【{r['name']}】{r['stock']} 買@{r['buy_price']} 停損:{r['stop_loss']} 目標:{r['target']} {r['status']}")

print("✅ 停損警報系統載入完成")
```

```python
# ===== AEGIS AI 機器學習模型 =====
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import StandardScaler
import numpy as np

def build_training_data(stock_list, period="2y"):
    """
    從歷史數據建立訓練集
    特徵：技術指標
    標籤：5天後是否上漲
    """
    all_data = []

    print(f"建立訓練資料，共 {len(stock_list)} 檔...")

    for stock_id in stock_list:
        try:
            df = yf.Ticker(stock_id).history(period=period)
            if df.empty or len(df) < 120:
                continue

            close  = df["Close"]
            volume = df["Volume"]

            # 特徵工程
            df["MA20"]  = close.rolling(20).mean()
            df["MA60"]  = close.rolling(60).mean()
            df["MA5"]   = close.rolling(5).mean()

            # RSI
            delta = close.diff()
            rs    = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
            df["RSI"] = 100 - (100 / (1 + rs))

            # MACD
            ema12 = close.ewm(span=12, adjust=False).mean()
            ema26 = close.ewm(span=26, adjust=False).mean()
            df["MACD"]   = ema12 - ema26
            df["Signal"] = df["MACD"].ewm(span=9, adjust=False).mean()
            df["MACD_hist"] = df["MACD"] - df["Signal"]

            # 布林通道位置
            bb_mid = close.rolling(20).mean()
            bb_std = close.rolling(20).std()
            df["BB_pos"] = (close - (bb_mid - 2*bb_std)) / (4*bb_std)

            # 成交量特徵
            df["Vol_ratio"] = volume / volume.rolling(20).mean()

            # 價格動能
            df["Mom5"]  = close.pct_change(5)
            df["Mom20"] = close.pct_change(20)

            # 均線距離
            df["MA20_dist"] = (close - df["MA20"]) / df["MA20"]
            df["MA60_dist"] = (close - df["MA60"]) / df["MA60"]

            # 標籤：5天後漲跌
            df["future_ret"] = close.shift(-5) / close - 1
            df["label"] = (df["future_ret"] > 0).astype(int)

            features = ["RSI", "MACD_hist", "BB_pos", "Vol_ratio",
                       "Mom5", "Mom20", "MA20_dist", "MA60_dist"]

            df = df.dropna(subset=features + ["label"])

            for _, row in df.iterrows():
                all_data.append({
                    "stock":      stock_id,
                    "RSI":        row["RSI"],
                    "MACD_hist":  row["MACD_hist"],
                    "BB_pos":     row["BB_pos"],
                    "Vol_ratio":  row["Vol_ratio"],
                    "Mom5":       row["Mom5"],
                    "Mom20":      row["Mom20"],
                    "MA20_dist":  row["MA20_dist"],
                    "MA60_dist":  row["MA60_dist"],
                    "label":      row["label"]
                })

            print(f"  {stock_id} ✅ {len(df)}筆")

        except Exception as e:
            print(f"  {stock_id} ❌ {e}")

    import pandas as pd
    result = pd.DataFrame(all_data)
    print(f"\n總訓練資料：{len(result)} 筆")
    return result

print("✅ 訓練資料建立函式載入完成")
```

```python
# ===== 訓練 AEGIS AI 機器學習模型 =====
feature_cols = ["RSI", "MACD_hist", "BB_pos", "Vol_ratio",
                "Mom5", "Mom20", "MA20_dist", "MA60_dist"]

X = train_df[feature_cols]
y = train_df["label"]

# 標準化
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 切分訓練/測試集
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

# 訓練兩個模型
print("訓練 Random Forest...")
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=20,
    random_state=42
)
rf_model.fit(X_train, y_train)
rf_acc = accuracy_score(y_test, rf_model.predict(X_test))

print("訓練 Gradient Boosting...")
gb_model = GradientBoostingClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    random_state=42
)
gb_model.fit(X_train, y_train)
gb_acc = accuracy_score(y_test, gb_model.predict(X_test))

print(f"\n===== AEGIS AI 模型訓練結果 =====")
print(f"Random Forest 準確率：{round(rf_acc*100, 1)} %")
print(f"Gradient Boosting 準確率：{round(gb_acc*100, 1)} %")

# 特徵重要性
importances = rf_model.feature_importances_
print(f"\n===== 最重要的指標 =====")
for feat, imp in sorted(zip(feature_cols, importances), key=lambda x: -x[1]):
    bar = "█" * int(imp * 50)
    print(f"{feat:<12} {bar} {round(imp*100, 1)}%")

# 選最好的模型
best_model = rf_model if rf_acc >= gb_acc else gb_model
best_name  = "Random Forest" if rf_acc >= gb_acc else "Gradient Boosting"
print(f"\n✅ 選用模型：{best_name}（{round(max(rf_acc,gb_acc)*100,1)}%）")
```

```python
def ml_predict_score(price_df):
    """
    用機器學習模型預測，回傳 0-100 分
    """
    try:
        if len(price_df) < 60:
            return 50  # 資料不足，給中性分

        close  = price_df["Close"]
        volume = price_df["Volume"]

        # 計算特徵
        ma20 = close.rolling(20).mean()
        ma60 = close.rolling(60).mean()
        ema12 = close.ewm(span=12, adjust=False).mean()
        ema26 = close.ewm(span=26, adjust=False).mean()
        macd  = ema12 - ema26
        signal = macd.ewm(span=9, adjust=False).mean()
        bb_mid = close.rolling(20).mean()
        bb_std = close.rolling(20).std()

        delta = close.diff()
        rs    = delta.clip(lower=0).rolling(14).mean() / (-delta.clip(upper=0)).rolling(14).mean()
        rsi   = 100 - (100 / (1 + rs))

        features = {
            "RSI":       rsi.iloc[-1],
            "MACD_hist": (macd - signal).iloc[-1],
            "BB_pos":    ((close - (bb_mid - 2*bb_std)) / (4*bb_std)).iloc[-1],
            "Vol_ratio": (volume / volume.rolling(20).mean()).iloc[-1],
            "Mom5":      close.pct_change(5).iloc[-1],
            "Mom20":     close.pct_change(20).iloc[-1],
            "MA20_dist": ((close - ma20) / ma20).iloc[-1],
            "MA60_dist": ((close - ma60) / ma60).iloc[-1],
        }

        X = np.array([[features[f] for f in feature_cols]])
        X_scaled = scaler.transform(X)

        # 預測機率
        prob = best_model.predict_proba(X_scaled)[0][1]  # 上漲機率

        # 轉換成 0-100 分
        score = round(prob * 100, 1)
        return score

    except Exception as e:
        return 50

# 測試
test_df = yf.Ticker("2330.TW").history(period="6mo")
ml_score = ml_predict_score(test_df)
print(f"===== AEGIS AI ML 預測 =====")
print(f"台積電 2330.TW")
print(f"ML 上漲機率：{ml_score}%")
if ml_score >= 60:
    print("AI 判定：偏多 📈")
elif ml_score <= 40:
    print("AI 判定：偏空 📉")
else:
    print("AI 判定：中性 ➡️")

print("✅ ML 預測函式載入完成")
```

```python
def aegis_full_score_v2(stock_id, keyword=None):
    """
    四維整合：基本面 + 技術面 + 新聞 + ML預測
    """
    try:
        ticker  = yf.Ticker(stock_id)
        info    = ticker.info
        price_df = ticker.history(period="6mo")

        if price_df.empty or len(price_df) < 60:
            return None

        # 基本面 (40%)
        fundamental = calculate_ai_score(info, price_df)

        # 技術面 (25%)
        technical, signals = get_technical_score(price_df)

        # ML預測 (20%)
        ml_score = ml_predict_score(price_df)

        # 新聞情緒 (15%)
        if keyword:
            news    = get_stock_news(keyword)
            news_score = min(max(50 + news["total_score"] * 10, 0), 100)
            sentiment  = news["sentiment"]
        else:
            news_score = 50
            sentiment  = "未抓取"

        # 加權總分
        total = round(
            fundamental * 0.40 +
            technical   * 0.25 +
            ml_score    * 0.20 +
            news_score  * 0.15,
            1
        )

        # AI 建議
        if total >= 75:
            advice = "強烈買進 🔥"
        elif total >= 65:
            advice = "偏多觀察 📈"
        elif total >= 50:
            advice = "中性觀察 ➡️"
        elif total >= 40:
            advice = "偏弱保守 📉"
        else:
            advice = "風險偏高 ⚠️"

        return {
            "stock":       stock_id,
            "name":        get_name(stock_id),
            "total":       total,
            "fundamental": fundamental,
            "technical":   technical,
            "ml_score":    ml_score,
            "news_score":  news_score,
            "sentiment":   sentiment,
            "signals":     signals,
            "advice":      advice
        }

    except Exception as e:
        print(f"{stock_id} 失敗：{e}")
        return None

print("✅ 四維整合評分 V2 載入完成")
```

```python
import warnings
warnings.filterwarnings("ignore")

def position_sizing(results, total_capital=1000000):
    """
    凱利公式資金管理
    根據 AI 分數自動分配倉位
    """
    print("===== AEGIS AI 資金管理系統 =====")
    print(f"總資金：{total_capital:,.0f} 元\n")

    # 只考慮分數 >= 60 的股票
    candidates = [r for r in results if r["total"] >= 60]

    if not candidates:
        print("目前沒有符合條件的股票")
        return

    # 根據分數加權分配
    total_score = sum(r["total"] for r in candidates)

    positions = []
    for r in candidates:
        weight    = r["total"] / total_score
        capital   = round(total_capital * weight, 0)
        risk_amt  = round(capital * 0.07, 0)  # 停損 7%

        positions.append({
            "stock":   r["stock"],
            "name":    r["name"],
            "score":   r["total"],
            "advice":  r["advice"],
            "weight":  round(weight * 100, 1),
            "capital": capital,
            "risk":    risk_amt,
        })

    print(f"{'股票':<10} {'名稱':<8} {'分數':<6} {'倉位%':<7} {'投入金額':<12} {'最大虧損'} {'建議'}")
    print("-" * 70)

    total_risk = 0
    for p in positions:
        print(f"{p['stock']:<10} {p['name']:<8} {p['score']:<6} {p['weight']:<7} {p['capital']:>10,.0f}  {p['risk']:>8,.0f}  {p['advice']}")
        total_risk += p["risk"]

    print(f"\n總風險暴露：{total_risk:,.0f} 元（{round(total_risk/total_capital*100,1)}%）")

    if total_risk / total_capital > 0.15:
        print("⚠️ 風險偏高，建議減少持股數量")
    else:
        print("✅ 風險控制良好")

    return positions

# 測試
positions = position_sizing(v2_results, total_capital=1000000)
```
